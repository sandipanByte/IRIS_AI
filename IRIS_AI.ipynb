```python
%%writefile /content/IRIS_AI/script.js

const $ = id => document.getElementById(id);

const video = $("video");
const canvas = $("canvas");

const ctx = canvas.getContext("2d");

let stream = null;
let running = false;
let paused = false;
let muted = false;
let starting = false;
let processing = false;
let session = 0;

let trails = {};
let previous = {};
let waveStates = {};

let candidates = {};
let announced = {};

let lastSpeech = 0;

const colors = [

    "#ff4757",
    "#00e5ff",
    "#00ff88",
    "#ffd700",
    "#c084fc"

];

const tips = [4, 8, 12, 16, 20];


// AUTOMATIC AI VOICE

function speak(text) {

    $("response").textContent = text;

    const entry = document.createElement("div");

    entry.textContent =
        new Date().toLocaleTimeString() + " — " + text;

    $("history").prepend(entry);

    while ($("history").children.length > 30) {

        $("history").lastChild.remove();

    }

    if (muted || paused || !window.speechSynthesis) return;

    const speech = new SpeechSynthesisUtterance(text);

    speech.rate = Number($("speed").value);

    speech.volume = Number($("volume").value);

    speech.pitch = 0.85;

    speech.lang = "en-GB";

    const voices = speechSynthesis.getVoices();

    const maleVoice = voices.find(v =>

        /Daniel|David|George|Ryan|Guy|Male/i.test(v.name)

    );

    if (maleVoice) {

        speech.voice = maleVoice;

    }

    speechSynthesis.cancel();

    speechSynthesis.speak(speech);

}


// START CAMERA

async function startCamera() {

    if (running || starting || processing) return;

    if (typeof Hands === "undefined") {

        $("response").textContent =
            "Hand recognition library unavailable. Check internet.";

        return;

    }

    starting = true;

    const current = ++session;

    try {

        if (!navigator.mediaDevices?.getUserMedia) {

            throw new Error("Open the website using HTTPS.");

        }

        const cameraStream =
            await navigator.mediaDevices.getUserMedia({

                video: {

                    width: { ideal: 640 },

                    height: { ideal: 480 }

                },

                audio: false

            });

        if (current !== session) {

            cameraStream.getTracks().forEach(t => t.stop());

            return;

        }

        stream = cameraStream;

        video.srcObject = stream;

        await video.play();

        if (current !== session) return;

        canvas.width = video.videoWidth;

        canvas.height = video.videoHeight;

        running = true;

        paused = false;

        $("pauseButton").textContent = "Pause AI";

        $("status").textContent = "● SYSTEM ONLINE";

        stream.getVideoTracks()[0].onended = stopCamera;

        speak("Hello. I am Iris. My visual intelligence system is online.");

        processFrame(current);

    } catch (error) {

        stopCamera();

        $("response").textContent =
            "Camera error: " + error.message;

    } finally {

        starting = false;

    }

}


// STOP CAMERA

function stopCamera() {

    running = false;

    session++;

    if (stream) {

        stream.getTracks().forEach(track => track.stop());

        stream = null;

    }

    video.srcObject = null;

    window.speechSynthesis?.cancel();

    clearTrails();

    candidates = {};

    announced = {};

    $("status").textContent = "● SYSTEM OFFLINE";

    $("gesture").textContent = "Waiting...";

    $("handCount").textContent = "0 Hands";

    $("movement").textContent = "No movement";

    $("confidence").textContent = "--";

    $("object").textContent =
        detector ? "Object recognition ready" : "Not enabled";

}


// PAUSE AND RESUME

function pauseAI() {

    if (!running) return;

    paused = !paused;

    candidates = {};

    window.speechSynthesis?.cancel();

    $("status").textContent = paused

        ? "● AI PAUSED"

        : "● SYSTEM ONLINE";

    $("pauseButton").textContent = paused

        ? "Resume AI"

        : "Pause AI";

    clearTrails();

}


// MUTE AND UNMUTE

function muteAI() {

    muted = !muted;

    window.speechSynthesis?.cancel();

    $("muteButton").textContent = muted

        ? "Unmute Voice"

        : "Mute Voice";

}


// CLEAR MOTION TRAILS

function clearTrails() {

    trails = {};

    previous = {};

    waveStates = {};

    previousHandDistance = null;

    ctx.clearRect(0, 0, canvas.width, canvas.height);

}


// RESET APPLICATION

function resetAI() {

    stopCamera();

    candidates = {};

    announced = {};

    lastSpeech = 0;

    paused = false;

    $("history").innerHTML = "";

    $("gesture").textContent = "Waiting...";

    $("movement").textContent = "No movement";

    $("response").textContent = "IRIS AI system reset.";

    $("pauseButton").textContent = "Pause AI";

}

window.addEventListener("pagehide", stopCamera);
```

```python
%%writefile -a /content/IRIS_AI/script.js

// DISTANCE BETWEEN LANDMARKS

function distance(a, b) {

    return Math.hypot(

        a.x - b.x,

        a.y - b.y,

        (a.z || 0) - (b.z || 0)

    );

}


// CALCULATE JOINT ANGLE

function angle(a, b, c) {

    const u = [a.x-b.x, a.y-b.y, a.z-b.z];

    const v = [c.x-b.x, c.y-b.y, c.z-b.z];

    const dot = u.reduce(

        (sum, value, i) => sum + value*v[i], 0

    );

    const length = Math.hypot(...u) * Math.hypot(...v);

    if (!length) return 0;

    return Math.acos(

        Math.max(-1, Math.min(1, dot/length))

    ) * 180 / Math.PI;

}


// RECOGNIZE HAND GESTURES

function recognizeGesture(lm) {

    const palm = distance(lm[0], lm[9]);

    const thumb =

        angle(lm[2], lm[3], lm[4]) > 145 &&

        distance(lm[4], lm[5]) > palm * 0.55;

    const fingers = [thumb];

    for (const tip of [8, 12, 16, 20]) {

        fingers.push(

            angle(lm[tip-3], lm[tip-2], lm[tip]) > 150 &&

            distance(lm[tip], lm[0]) >

            distance(lm[tip-2], lm[0])

        );

    }

    const [t, i, m, r, p] = fingers;

    const count = fingers.filter(Boolean).length;


    // OK SIGN

    if (

        distance(lm[4], lm[8]) < palm * 0.3 &&

        m && r && p

    ) {

        return "OK Sign";

    }


    // THUMBS UP AND DOWN

    if (t && !i && !m && !r && !p) {

        if (lm[4].y < lm[2].y - palm * 0.3) {

            return "Thumbs Up";

        }

        if (lm[4].y > lm[2].y + palm * 0.3) {

            return "Thumbs Down";

        }

    }


    // PEACE SIGN

    if (!t && i && m && !r && !p) {

        return "Peace Sign";

    }


    // ROCK SIGN

    if (!t && i && !m && !r && p) {

        return "Rock Sign";

    }


    // OPEN PALM

    if (count === 5) {

        return "Open Palm";

    }


    // CLOSED FIST

    if (count === 0) {

        return "Closed Fist";

    }


    // POINTING FINGER

    if (!t && i && !m && !r && !p) {

        return "Pointing Finger";

    }


    return count + (count === 1 ? " Finger" : " Fingers");

}


// INTELLIGENT VOICE RESPONSES

const replies = {

    "Thumbs Up":
        "Hey! I see a thumbs up. Nice!",

    "Thumbs Down":
        "I can see your thumb pointing down.",

    "Peace Sign":
        "Peace! You are showing two fingers.",

    "Open Palm":
        "I can see your open palm and all five fingers.",

    "Closed Fist":
        "You just closed your fist.",

    "Pointing Finger":
        "I see your pointing finger.",

    "OK Sign":
        "That looks like an OK gesture.",

    "Rock Sign":
        "Nice! I recognize that rock sign.",

    "1 Finger":
        "You are showing one finger.",

    "2 Fingers":
        "You are showing two fingers.",

    "3 Fingers":
        "I can see three fingers.",

    "4 Fingers":
        "I can see four fingers.",

    "Waving":
        "Hello! I see you waving at me.",

    "Both Hands":
        "I can see both of your hands.",

    "Hands Moving Together":
        "It looks like your hands are moving together."

};


// PREVENT REPETITIVE RESPONSES

function meaningfulEvent(channel, event, message) {

    if (!event) {

        delete candidates[channel];

        delete announced[channel];

        return;

    }

    const now = Date.now();

    if (candidates[channel]?.event !== event) {

        candidates[channel] = {

            event: event,

            since: now

        };

        return;

    }

    // Gesture must remain stable

    if (now - candidates[channel].since < 450) return;

    // Avoid repeating the current gesture

    if (announced[channel] === event) return;

    // Voice cooldown

    if (now - lastSpeech < 2200) return;

    // Prevent overlapping voices

    if (window.speechSynthesis?.speaking && !muted) return;

    announced[channel] = event;

    lastSpeech = now;

    speak(

        message ||

        replies[event] ||

        `I can see ${event}.`

    );

}
```

```python
%%writefile -a /content/IRIS_AI/script.js

// HAND JOINT CONNECTIONS

const connections = [

    [0,1], [1,2], [2,3], [3,4],

    [0,5], [5,6], [6,7], [7,8],

    [5,9], [9,10], [10,11], [11,12],

    [9,13], [13,14], [14,15], [15,16],

    [13,17], [17,18], [18,19], [19,20],

    [0,17]

];


// INITIALIZE HAND RECOGNITION

let hands = null;

function initializeHands() {

    if (hands) return;

    hands = new Hands({

        locateFile: file =>

            `https://cdn.jsdelivr.net/npm/@mediapipe/hands@0.4.1675469240/${file}`

    });

    hands.setOptions({

        maxNumHands: 2,

        modelComplexity: 0,

        minDetectionConfidence: 0.65,

        minTrackingConfidence: 0.65

    });

    hands.onResults(handleResults);

}


// DRAW HAND LANDMARKS

function drawHand(lm, id) {

    const w = canvas.width;

    const h = canvas.height;


    // CONNECT HAND JOINTS

    for (const [a,b] of connections) {

        ctx.beginPath();

        ctx.moveTo(lm[a].x*w, lm[a].y*h);

        ctx.lineTo(lm[b].x*w, lm[b].y*h);

        ctx.strokeStyle = "#00e5ff";

        ctx.lineWidth = 2;

        ctx.stroke();

    }


    // DRAW ALL 21 LANDMARKS

    lm.forEach((point, index) => {

        const finger = Math.min(

            4,

            Math.max(0, Math.floor((index-1)/4))

        );

        ctx.beginPath();

        ctx.arc(

            point.x*w,

            point.y*h,

            4,

            0,

            Math.PI*2

        );

        ctx.fillStyle = index === 0

            ? "white"

            : colors[finger];

        ctx.fill();

    });


    // SAVE INDIVIDUAL FINGER MOVEMENT

    tips.forEach((tip, index) => {

        const key = id + "-" + tip;

        if (!trails[key]) {

            trails[key] = [];

        }

        trails[key].push({

            x: lm[tip].x*w,

            y: lm[tip].y*h,

            time: performance.now(),

            color: colors[index]

        });

        const limit = Number($("trailLength").value);

        while (trails[key].length > limit) {

            trails[key].shift();

        }

    });

}


// DRAW FADING MOTION TRAILS

function drawTrails() {

    const now = performance.now();

    const lifetime = Number($("trailLength").value) * 80;

    for (const key of Object.keys(trails)) {

        trails[key] = trails[key].filter(

            point => now - point.time < lifetime

        );

        const points = trails[key];

        for (let j = 1; j < points.length; j++) {

            const opacity = Math.max(

                0,

                1 - (now - points[j].time) / lifetime

            );

            ctx.globalAlpha = opacity * 0.85;

            ctx.beginPath();

            ctx.moveTo(points[j-1].x, points[j-1].y);

            ctx.lineTo(points[j].x, points[j].y);

            ctx.strokeStyle = points[j].color;

            ctx.lineWidth = 3;

            ctx.stroke();

        }

        if (!points.length) {

            delete trails[key];

        }

    }

    ctx.globalAlpha = 1;

}
```

```python
%%writefile -a /content/IRIS_AI/script.js

let previousHandDistance = null;


// RECOGNIZE MOVEMENTS

function recognizeMovement(lm, id, gesture) {

    const current = lm[0];

    const old = previous[id];

    previous[id] = {

        x: current.x,

        y: current.y

    };

    if (!old) return "Hand detected";

    const dx = current.x - old.x;

    const dy = current.y - old.y;


    // WAVING DETECTION

    if (!waveStates[id]) {

        waveStates[id] = {

            direction: 0,

            turns: [],

            activeUntil: 0

        };

    }

    const wave = waveStates[id];

    const now = Date.now();

    if (gesture === "Open Palm" && Math.abs(dx) > 0.012) {

        const direction = Math.sign(dx);

        if (wave.direction && direction !== wave.direction) {

            wave.turns.push(now);

        }

        wave.direction = direction;

    }

    wave.turns = wave.turns.filter(

        t => now - t < 1800

    );

    if (wave.turns.length >= 3) {

        wave.activeUntil = now + 1000;

    }

    if (gesture === "Open Palm" && now < wave.activeUntil) {

        return "Waving";

    }


    // DIRECTION DETECTION

    if (Math.abs(dx) > Math.abs(dy)) {

        if (dx > 0.018) return "Moving Left";

        if (dx < -0.018) return "Moving Right";

    }

    if (dy > 0.018) return "Moving Down";

    if (dy < -0.018) return "Moving Up";

    return "Hand Stable";

}


// ANALYZE EACH CAMERA FRAME

function handleResults(results) {

    if (!running || paused) return;

    ctx.clearRect(0, 0, canvas.width, canvas.height);

    const detected = results.multiHandLandmarks || [];

    $("handCount").textContent =

        detected.length + " Hand(s)";

    const gestures = [];

    const movements = [];

    const activeIds = new Set();


    detected.forEach((lm, index) => {

        const label = results.multiHandedness?.[index]?.label;

        const id = label && !activeIds.has(label)

            ? label

            : String(index);

        activeIds.add(id);

        drawHand(lm, id);

        const gesture = recognizeGesture(lm);

        const movement = recognizeMovement(lm, id, gesture);

        gestures.push(gesture);

        movements.push(movement);

    });


    // REMOVE OLD HAND STATES

    for (const id of Object.keys(previous)) {

        if (!activeIds.has(id)) {

            delete previous[id];

            delete waveStates[id];

        }

    }

    drawTrails();


    // NO HAND DETECTED

    if (!detected.length) {

        $("gesture").textContent = "No hand detected";

        $("movement").textContent = "No movement";

        previousHandDistance = null;

        meaningfulEvent("gesture", null);

        meaningfulEvent("action", null);

        return;

    }


    // RECOGNIZE COMBINED ACTIONS

    let action = "";

    if (movements.includes("Waving")) {

        action = "Waving";

    } else if (detected.length === 2) {

        const gap = distance(

            detected[0][0],

            detected[1][0]

        );

        action = previousHandDistance !== null &&

                 gap < previousHandDistance - 0.015

            ? "Hands Moving Together"

            : "Both Hands";

        previousHandDistance = gap;

    } else {

        previousHandDistance = null;

    }


    // UPDATE DASHBOARD

    $("gesture").textContent = gestures.join(" | ");

    $("movement").textContent =

        action || movements.join(" | ");

    meaningfulEvent("action", action);

    if (action === "Waving") {

        meaningfulEvent("gesture", null);

    } else {

        meaningfulEvent("gesture", gestures[0]);

    }

}


// CONTINUOUS CAMERA PROCESSING

async function processFrame(current) {

    if (!running || current !== session) return;

    if (!paused && video.readyState >= 2) {

        processing = true;

        try {

            initializeHands();

            await hands.send({

                image: video

            });

            if (running && current === session) {

                await detectObjects(current);

            }

        } catch (error) {

            stopCamera();

            $("response").textContent =

                "Recognition error: " + error.message;

        } finally {

            processing = false;

        }

    }

    if (running && current === session) {

        setTimeout(

            () => processFrame(current),

            65

        );

    }

}
```

```python
%%writefile -a /content/IRIS_AI/script.js

let detector = null;

let loadingObjects = false;

let lastObjectCheck = 0;


// LOAD BROWSER LIBRARY

function loadScript(url) {

    return new Promise((resolve, reject) => {

        const script = document.createElement("script");

        script.src = url;

        script.onload = resolve;

        script.onerror = () => {

            script.remove();

            reject(new Error("Unable to load recognition library."));

        };

        document.head.appendChild(script);

    });

}


// ENABLE OBJECT RECOGNITION

async function enableObjects() {

    if (detector || loadingObjects) return;

    loadingObjects = true;

    $("object").textContent = "Loading object recognition...";

    try {

        if (!window.tf) {

            await loadScript(

                "https://cdn.jsdelivr.net/npm/@tensorflow/tfjs@4.22.0/dist/tf.min.js"

            );

        }

        if (!window.cocoSsd) {

            await loadScript(

                "https://cdn.jsdelivr.net/npm/@tensorflow-models/coco-ssd@2.2.2/dist/coco-ssd.min.js"

            );

        }

        detector = await cocoSsd.load({

            base: "lite_mobilenet_v2"

        });

        $("object").textContent = "Object recognition ready";

        if (running && !paused) {

            speak("Object recognition is now available.");

        }

    } catch (error) {

        $("object").textContent =

            "Unable to load. Check your internet connection.";

    } finally {

        loadingObjects = false;

    }

}


// OBJECT DESCRIPTIONS

const descriptions = {

    "cell phone":
        "That looks like a smartphone. It is a portable communication device.",

    "book":
        "I can see a book. It contains written or printed information.",

    "bottle":
        "You appear to be showing me a bottle.",

    "cup":
        "That looks like a cup, commonly used for drinking.",

    "laptop":
        "I can see a laptop. It is a portable computer.",

    "keyboard":
        "That appears to be a computer keyboard.",

    "mouse":
        "I can see a computer mouse.",

    "apple":
        "That looks like an apple.",

    "banana":
        "I can see a banana.",

    "clock":
        "That appears to be a clock.",

    "chair":
        "I can see a chair.",

    "backpack":
        "That looks like a backpack.",

    "sports ball":
        "I can see what appears to be a sports ball.",

    "teddy bear":
        "That looks like a teddy bear.",

    "remote":
        "I can see a remote control."

};


// ANALYZE OBJECTS

async function detectObjects(current) {

    if (!detector || paused || !running) return;

    if (Date.now() - lastObjectCheck < 1200) return;

    lastObjectCheck = Date.now();

    try {

        const predictions = await detector.detect(

            video,

            10,

            0.65

        );

        if (!running || paused || current !== session) return;

        const objects = predictions

            .filter(p =>

                p.class !== "person" && p.score >= 0.65

            )

            .sort((a,b) => b.score - a.score);

        if (!objects.length) {

            $("object").textContent = "No recognized object";

            $("confidence").textContent = "--";

            meaningfulEvent("object", null);

            return;

        }

        const item = objects[0];

        const confidence = Math.round(item.score * 100);

        $("object").textContent = item.class.toUpperCase();

        $("confidence").textContent = confidence + "%";

        meaningfulEvent(

            "object",

            item.class,

            descriptions[item.class] ||

            `That looks like ${item.class}.`

        );

    } catch (error) {

        $("object").textContent = "Object recognition unavailable";

        $("confidence").textContent = "--";

        meaningfulEvent("object", null);

    }

}
```
